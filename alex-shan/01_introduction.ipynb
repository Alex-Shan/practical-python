## 1.5 列表

```python
symbols = 'HPQ,AAPL,IBM,MSFT,YHOO,DOA,GOOG'
symlist = symbols.split(',')
# 练习 1.19：提取和重新分配列表元素
print(symlist[0])
symlist[2] = 'AIG'
print(symlist)

# 练习 1.20：遍历列表元素
for s in symlist:
    print('s =', s)

# 练习 1.21：成员测试
print(f"'IBM' in symlist = {'IBM' in symlist}")

# 练习 1.22：添加，插入和删除元素
print('symlist =', symlist)
symlist.append('SCO')
print('symlist.append(SCO) =', symlist)
symlist.insert(2, 'HPQ')
print('symlist.insert(2, HPQ) =', symlist)
del symlist[2]
print('del symlist[2] =', symlist)

# 练习 1.23：排序
symlist.sort()
print('symlist.sort() =', symlist)

# 练习 1.24：列表转字符串
symbols_string = ','.join(symlist)
print(f"','.join(symlist) = {symbols_string}")
symbols_string = ''.join(symlist)
print(f"''.join(symlist) = {symbols_string}")

# 练习 1.25：包含任何内容的列表
nums = [101, 102, 103]
items = ['spam', symlist, nums]
print('items =', items)
```

## 1.7 函数

```python
# 练习 1.29：定义一个函数
def greeting(name):
    'Issues a greeting'
    print('Hello', name)

greeting('Guido')

# 练习 1.30：将脚本转换为函数
def portfolio_cost(filename):
    total_cost = 0.0
    with open(filename, 'rt') as f:
        next(f)  # 跳过表头: name,shares,price
        for line in f:
            name, shares, price = line.split(',')
            total_cost += int(shares) * float(price)
    return total_cost

cost = portfolio_cost('../Work/Data/portfolio.csv')
print('Total cost:', cost)

# 练习 1.31：错误处理(消除错误行)
try:
    cost = portfolio_cost('../Work/Data/missing.csv')
    print('Total cost:', cost)
except Exception as e:
    print('Error:', e)

# 练习 1.32：使用库函数

import csv
f = open('../Work/Data/portfolio.csv')
rows = csv.reader(f)
headers = next(rows)
print(headers)
for row in rows:
        print(row)
f.close()
```
