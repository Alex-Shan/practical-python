## 3.1 脚本 

```python
# report.py
import csv

def read_portfolio(filename):
    '''
    Read a stock portfolio file into a list of dictionaries with keys
    name, shares, and price.
    '''
    portfolio = []
    with open(filename) as f:
        rows = csv.reader(f)
        headers = next(rows)

        for row in rows:
            record = dict(zip(headers, row))
            stock = {
                'name' : record['name'],
                'shares' : int(record['shares']),
                'price' : float(record['price'])
            }
            portfolio.append(stock)
    return portfolio

# 练习 3.1：将程序构造为函数的集合

def print_report(report):
    '''
    Print a nicely formatted report of a stock portfolio.
    '''
    print('%10s %10s %10s %10s' % ('Name', 'Shares', 'Price', 'Value'))
    print(('-' * 10 + ' ') * 4)
    for stock in report:
        value = stock['shares'] * stock['price']
        print('%10s %10d %10.2f %10.2f' %
              (stock['name'], stock['shares'], stock['price'], value))


# 练习 3.2：为程序执行创建一个顶层函数
def portfolio_report(portfolio_file, prices_file):
    '''
    Generate a report of a stock portfolio.
    '''
    portfolio = read_portfolio(portfolio_file)
    print_report(portfolio)


portfolio_report('../Work/Data/portfolio2.csv', '../Work/Data/prices.csv')

files = ['../Work/Data/portfolio.csv', '../Work/Data/portfolio2.csv']
for name in files:
    print(f'{name:-^43s}')
    portfolio_report(name, '../Work/Data/prices.csv')
    print()
    
```

```python
# 练习 3.3：读取 CSV 文件

import csv

def parse_csv(filename):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)

        # Read the file headers
        headers = next(rows)
        records = []
        for row in rows:
            if not row:    # Skip rows with no data
                continue
            record = dict(zip(headers, row))
            records.append(record)

    return records

portfolio = parse_csv('../Work/Data/portfolio.csv')
print(portfolio)

# 练习 3.4：构建列选择器

def parse_csv(filename,select=None):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)

        # Read the file headers
        headers = next(rows)
        records = []
        for row in rows:
            if not row:    # Skip rows with no data
                continue
            record = dict(zip(headers, row))
            if select:
                record = {key: record[key] for key in select}
            records.append(record)

    return records

shares_held = parse_csv('../Work/Data/portfolio.csv', select=['name','shares'])
print(shares_held)

# 练习 3.5：执行类型转换
def parse_csv(filename,select=None,types=None):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)

        # Read the file headers
        headers = next(rows)
        records = []
        for row in rows:
            if not row:    # Skip rows with no data
                continue
            record = dict(zip(headers, row))
            if select:
                record = {key: record[key] for key in select}
            if types:
                record = {key: func(record[key]) for key, func in zip(record.keys(), types)}
            records.append(record)

    return records

portfolio = parse_csv('../Work/Data/portfolio.csv', types=[str, int, float])
shares_held = parse_csv('../Work/Data/portfolio.csv', select=['name', 'shares'], types=[str, int])
print(portfolio)
print(shares_held)


# 练习 3.6：处理无标题的数据
def parse_csv(filename,select=None,types=None,has_headers=True):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)

        # Read the file headers
        if has_headers:
            headers = next(rows)
        records = []
        for row in rows:
            if not row:    # Skip rows with no data
                continue
            if has_headers:
                record = dict(zip(headers, row))
                if select:
                    record = {key: record[key] for key in select}
                if types:
                    record = {key: func(record[key]) for key, func in zip(record.keys(), types)}
            else:
                record = tuple(row)
                if select:
                    record = tuple(record[i] for i in select)
                if types:
                    record = tuple(func(val) for func, val in zip(types, record))

            records.append(record)

    return records

prices = parse_csv('../Work/Data/prices.csv', types=[str,float], has_headers=False)
print(prices)


# 练习 3.7：选择其它的列分隔符
def parse_csv(filename,select=None,types=None,has_headers=True,delimiter=','):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f,delimiter=delimiter)

        # Read the file headers
        if has_headers:
            headers = next(rows)
        records = []
        for row in rows:
            if not row:    # Skip rows with no data
                continue
            if has_headers:
                record = dict(zip(headers, row))
                if select:
                    record = {key: record[key] for key in select}
                if types:
                    record = {key: func(record[key]) for key, func in zip(record.keys(), types)}
            else:
                record = tuple(row)
                if select:
                    record = tuple(record[i] for i in select)
                if types:
                    record = tuple(func(val) for func, val in zip(types, record))

            records.append(record)

    return records
portfolio = parse_csv('../Work/Data/portfolio.dat', types=[str, int, float], delimiter=' ')
print(portfolio)


```

```python
# 练习 3.8：触发异常

def parse_csv(filename,select=None,types=None,has_headers=True):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)
        if select and not has_headers:
            raise RuntimeError("select argument requires column headers")

        # Read the file headers
        if has_headers:
            headers = next(rows)
        records = []
        for row in rows:
            if not row:    # Skip rows with no data
                continue
            if has_headers:
                record = dict(zip(headers, row))
                if select:
                    record = {key: record[key] for key in select}
                if types:
                    record = {key: func(record[key]) for key, func in zip(record.keys(), types)}
            else:
                record = tuple(row)
                if types:
                    record = tuple(func(val) for func, val in zip(types, record))

            records.append(record)

    return records

try:
    parse_csv(
        '../Work/Data/prices.csv',
        select=['name', 'price'],
        has_headers=False
    )
except Exception as e:
    print(f'Error: {e}')

# 练习 3.9：捕获异常

def parse_csv(filename,select=None,types=None,has_headers=True):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)
        if select and not has_headers:
            raise RuntimeError("select argument requires column headers")

        # Read the file headers
        if has_headers:
            headers = next(rows)
        records = []
        for rowno, row in enumerate(rows, start=1):
            if not row:
                continue

            # 先建立记录，再选择需要的列
            if has_headers:
                record = dict(zip(headers, row))
                if select:
                    record = {key: record[key] for key in select}
            else:
                record = tuple(row)

            # 尝试进行类型转换
            try:
                if types:
                    if has_headers:
                        record = {
                            key: func(value)
                            for (key, value), func
                            in zip(record.items(), types)
                        }
                    else:
                        record = tuple(
                            func(value)
                            for func, value in zip(types, record)
                        )
            except ValueError as e:
                print(f"Row {rowno}: Couldn't convert {row}")
                print(f"Row {rowno}: Reason {e}")
                continue

            records.append(record)

    return records

portfolio = parse_csv(
    '../Work/Data/missing.csv',
    types=[str, int, float]
)
print(portfolio)

# 练习 3.10：隐藏错误

def parse_csv(filename, select=None, types=None,
              has_headers=True, silence_errors=False):
    '''
    Parse a CSV file into a list of records
    '''
    with open(filename) as f:
        rows = csv.reader(f)
        if select and not has_headers:
            raise RuntimeError("select argument requires column headers")

        # Read the file headers
        if has_headers:
            headers = next(rows)
        records = []
        for rowno, row in enumerate(rows, start=1):
            if not row:
                continue

            # 先建立记录，再选择需要的列
            if has_headers:
                record = dict(zip(headers, row))
                if select:
                    record = {key: record[key] for key in select}
            else:
                record = tuple(row)

            # 尝试进行类型转换
            try:
                if types:
                    if has_headers:
                        record = {
                            key: func(value)
                            for (key, value), func
                            in zip(record.items(), types)
                        }
                    else:
                        record = tuple(
                            func(value)
                            for func, value in zip(types, record)
                        )
            except ValueError as e:
                if not silence_errors:
                    print(f"Row {rowno}: Couldn't convert {row}")
                    print(f"Row {rowno}: Reason {e}")
                continue

            records.append(record)

    return records

# 隐藏错误消息
portfolio = parse_csv(
    '../Work/Data/missing.csv',
    types=[str, int, float],
    silence_errors=True
)
print(portfolio)
```
