## 2.2 容器

```python
# 练习 2.4：包含元组的列表
import csv

def read_portfolio(filename):
    portfolio = []
    with open(filename, 'rt') as f:
        rows = csv.reader(f)
        headers = next(rows)
        for row in rows:
            holding = (row[0], int(row[1]), float(row[2]))
            portfolio.append(holding)
    return portfolio

portfolio = read_portfolio('../Work/Data/portfolio.csv')
print(portfolio)

total = 0.0
for name, shares, price in portfolio:
    total += shares*price

print(total)

# 练习 2.5：包含字典的列表

def read_portfolio(filename):
    portfolio = []
    with open(filename, 'rt') as f:
        rows = csv.reader(f)
        headers = next(rows)
        for row in rows:
            holding = {'name': row[0], 'shares': int(row[1]), 'price': float(row[2])}
            portfolio.append(holding)
    return portfolio

portfolio = read_portfolio('../Work/Data/portfolio.csv')
print(portfolio)
for s in portfolio:
    total += s['shares']*s['price']

print(total)
```

## 2.3 格式化

```python
# 练习 2.8：如何格式化数字
value = 42863.1
print(value)
print(f'{value:0.4f}')
print(f'{value:>16.2f}')
print(f'{value:<16.2f}')
print(f'{value:*>16,.2f}')

# 练习 2.9：收集数据
def read_prices(filename):
    prices = {}
    with open(filename, 'rt') as f:
        rows = csv.reader(f)
        for row in rows:
            try:
                prices[row[0]] = float(row[1])
            except IndexError:
                pass
    return prices

def make_report(portfolio, prices):
    report = []
    for holding in portfolio:
        name = holding['name']
        shares = holding['shares']
        price = holding['price']
        current_price = prices[name]
        change = current_price - price
        report.append((name, shares, price, current_price, change))
    return report

portfolio = read_portfolio('../Work/Data/portfolio.csv')
prices = read_prices('../Work/Data/prices.csv')
report = make_report(portfolio, prices)
for r in report:
        print(r)


# 练习 2.10：打印格式化的表格
report = make_report(portfolio, prices)
for name, shares, price, current_price, change in report:
    print(f'{name:>10s} {shares:>10d} {price:>10.2f} {change:>10.2f}')

# 练习 2.11：添加标题
print()
print(f'{"Name":>10s} {"Shares":>10s} {"Price":>10s} {"Change":>10s}')
print(f'{"-"*10} {"-"*10} {"-"*10} {"-"*10}')
for name, shares, price, current_price, change in report:
    print(f'{name:>10s} {shares:>10d} {price:>10.2f} {change:>10.2f}')

# 练习 2.12：格式化挑战
print()
print(f'{"Name":>10s} {"Shares":>10s} {"Price":>10s} {"Change":>10s}')
print(f'{"-"*10} {"-"*10} {"-"*10} {"-"*10}')
for name, shares, price, current_price, change in report:
    price_str = f'¥{price:.2f}'
    print(f'{name:>10s} {shares:>10d} {price_str:>10s} {change:>10.2f}')
```
